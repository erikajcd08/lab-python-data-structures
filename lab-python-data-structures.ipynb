````python
# Paso 1
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# Paso 2
inventory = {}

# Paso 3
for product in products:
    inventory[product] = int(input(f"Introduce la cantidad de {product}: "))

# Paso 4
customer_orders = set()

# Paso 5 - Mejorando while
answer = "yes"
while answer == "yes":
    product = input("Introduce el nombre del producto: ")
    if product in products:
            customer_orders.add(product)
    else:
         print("Producto no válido, inténtalo de nuevo por favor")
    answer = input("¿Quieres añadir otro producto? (yes/no): ")

# Paso 6
print(customer_orders)

# Paso 7
total_products_ordered = len(customer_orders)
percentage = (len(customer_orders) / len(products)) * 100
order_status = (total_products_ordered, percentage)

# Paso 8
print("Order Statistics:")
print(f"Total Products Ordered: {order_status[0]}")
print(f"Percentage of Products Ordered: {order_status[1]}%")

# Paso 9 y 10 - Mejorado, solo productos pedidos
for product in customer_orders:
    inventory[product] -= 1

for product in inventory:
    print(f"{product}: {inventory[product]}")
````
